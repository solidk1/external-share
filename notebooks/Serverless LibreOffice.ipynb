```python
import stanza, os

# Skip model download if already cached
model_dir = os.path.expanduser("~/stanza_resources/zh")
if not os.path.exists(model_dir):
    print("Downloading Chinese model...")
    stanza.download('zh')
else:
    print("Chinese model already cached — skipping download.")

nlp = stanza.Pipeline(lang='zh', processors='tokenize')

text = "自然语言处理非常重要。Stanza 是一个强大的工具。"
doc = nlp(text)

for i, sentence in enumerate(doc.sentences):
    print(f"第 {i+1} 句: {sentence.text}")
    print("包含的词语: ")
    for token in sentence.tokens:
        print(f"[{token.text}]", end=" ")
    print("\n")
```

```python
%sh
set -uo pipefail

LIBS_PATH="$HOME/apps/libs/usr/lib/x86_64-linux-gnu"
export LD_LIBRARY_PATH="${LIBS_PATH}:${LD_LIBRARY_PATH:-}"

# Skip install if soffice already works
SOFFICE=$(find ~/apps/libreoffice/opt -name soffice -path '*/program/soffice' 2>/dev/null | head -1)
if [[ -n "$SOFFICE" ]] && $SOFFICE --version &>/dev/null; then
  echo "LibreOffice already installed — skipping."
  echo "soffice path: $SOFFICE"
  $SOFFICE --version
  exit 0
fi

# Resolve latest stable version (only when install is needed)
echo "Resolving latest LibreOffice version..."
VERSION=$(wget -qO- https://download.documentfoundation.org/libreoffice/stable/ \
  | grep -oP '\d+\.\d+\.\d+(?=/)' | sort -Vu | tail -1) \
  || { echo "ERROR: Failed to fetch version list" >&2; exit 1; }

if [[ ! "$VERSION" =~ ^[0-9]+\.[0-9]+\.[0-9]+$ ]]; then
  echo "ERROR: Failed to parse version (got: '$VERSION')" >&2
  exit 1
fi

echo "Installing LibreOffice $VERSION ..."
mkdir -p ~/apps && cd ~/apps

FILE="LibreOffice_${VERSION}_Linux_x86-64_deb.tar.gz"
wget -q "https://download.documentfoundation.org/libreoffice/stable/${VERSION}/deb/x86_64/${FILE}" \
  || { echo "ERROR: Download failed" >&2; exit 1; }
tar -xzf "$FILE"
rm "$FILE"

mkdir -p ~/apps/libreoffice
for pkg in LibreOffice_*/DEBS/*.deb; do
  dpkg-deb -x "$pkg" ~/apps/libreoffice
done
rm -rf LibreOffice_*/

mkdir -p ~/apps/libs
wget -q "http://archive.ubuntu.com/ubuntu/pool/main/libx/libxinerama/libxinerama1_1.1.4-3_amd64.deb" -O /tmp/libxinerama1.deb
dpkg-deb -x /tmp/libxinerama1.deb ~/apps/libs/
rm /tmp/libxinerama1.deb

SOFFICE=$(find ~/apps/libreoffice/opt -name soffice -path '*/program/soffice' | head -1)
echo "soffice path: $SOFFICE"
$SOFFICE --version
```

```python
import subprocess, glob, os

soffice_matches = glob.glob(os.path.expanduser("~/apps/libreoffice/opt/libreoffice*/program/soffice"))
if not soffice_matches:
    raise FileNotFoundError("LibreOffice not found — run the install cell first.")
SOFFICE = soffice_matches[0]

env = os.environ.copy()
libs_path = os.path.expanduser("~/apps/libs/usr/lib/x86_64-linux-gnu")
env["LD_LIBRARY_PATH"] = libs_path + ":" + env.get("LD_LIBRARY_PATH", "")

input_file = "/Volumes/shao_sandbox1/aaa.doc"
if not os.path.exists(input_file):
    raise FileNotFoundError(f"Input file not found: {os.path.abspath(input_file)}")

result = subprocess.run([
    SOFFICE,
    "--headless",
    "--convert-to", "pdf",
    "--outdir", ".",
    input_file
], capture_output=True, text=True, env=env)

print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError(f"LibreOffice conversion failed (exit {result.returncode}): {result.stderr}")
```
